# Atividade Prática 3 - Pandas para análise de dados do Titanic

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
CSV_PATH = 'Data/titanic.csv'

df = pd.read_csv(CSV_PATH, sep=',')
```

```python
df.head()
```

## Distribuição de Idade por Gênero e Classe Social

### Barplots de Estatísticas Descritivas Básicas (Média, Mediana e Desvio Padrão)

**Média de Idade por Classe Social/Sexo**

```python
df_mean_age = df.groupby(['Sex', 'Pclass'])['Age'].mean().unstack('Sex')

fig, ax = plt.subplots()
df_mean_age.plot(ax=ax, kind='bar', title='Média de idade em cada classe/sexo', ylabel='Idade', xlabel='Classe social')
plt.legend()
plt.show()

print(df_mean_age)
```

**Mediana da Idade por Classe Social/Sexo**

```python
df_median_age = df.groupby(['Sex', 'Pclass'])['Age'].median().unstack('Sex')

fig, ax = plt.subplots()
df_median_age.plot(ax=ax, kind='bar', title='Mediana da idade em cada classe/sexo', ylabel='Idade', xlabel='Classe social')
plt.legend()
plt.show()

print(df_median_age)
```

**Desvio Padrão por Classe Social/Sexo**

```python
df_std_age = df.groupby(['Sex', 'Pclass'])['Age'].std().unstack('Sex')

fig, ax = plt.subplots()
df_std_age.plot(ax=ax, kind='bar', title='Desvio padrão da idade em cada classe/sexo', ylabel='Idade', xlabel='Classe social')
plt.legend()

plt.show()
print(df_std_age)
```

## Taxa de Sobrevivência por Faixa Etária

**Adiciona atributo faixa etária (Crianças, Jovens, Adultos, Idosos)**

Crianças: 0 a 11 anos  
Jovens: 12 a 29 anos  
Adultos: 30 a 59 anos  
Idosos: 60 anos ou mais  

```python
limites = [0, 11, 29, 59, 200]
faixas_etarias = ['Criança', 'Jovem', 'Adulto', 'Idoso']

df['faixa_etaria'] = pd.cut(df['Age'], bins=limites, labels=faixas_etarias)

df.head()
```
